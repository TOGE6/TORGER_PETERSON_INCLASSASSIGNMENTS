```python
#garage fee
def calcFee(hours):
  fee = 5 + 2.5 * hours
  if fee < 10:
    fee = 10.0
  elif fee > 20:
    fee = 20.0
  print(f'Fee: {fee}$')

def main():
  try:
    numHours = int(input('Enter number of hours: '))
    calcFee(numHours)
  except ValueError:
    print('Cannot cast into int, Please enter a number.')
  except:
    print("unknown error...")
if __name__ == '__main__':
  main()
```

```python
#movie
print("Ad Spending ($)  Profit ($)")

ticket_price = 10
base_attendance = 20
fixed_costs = 200

for ad_spending in range(0, 201, 25):
    additional_attendees = 2 * round(ad_spending)** .5
    total_attendees = base_attendance + additional_attendees
    revenue = total_attendees * ticket_price
    profit = revenue - (ad_spending + fixed_costs)
    print(f"{ad_spending:<15}  {profit:<10}")
```

```python
class Greeting:
  def __init__(self, first, last):
    self.first = first
    self.last = last
  def display(self, formal=False):
    if formal == True:
      print(f'Hello, {self.first} {self.last}')
    else:
      print(f'Hi {self.first}')

  def __str__(self):
    #return str(self.first) + ", " + str(self.last)
    return str(self.first) + ", " + str(self.last)
    return f' {self.first}, {self.last}'

greeting = Greeting('Donald ', 'James')
other_greeting = Greeting('McDonald', 'Hank')
greeting.display()
print(greeting)
print(other_greeting)
```

```python
class Student:
    def __init__(self, name, number):
        self.name = name
        self.number = number
        self.courses = []
    def enroll(self, course):
        if course in self.courses:
            print(f'already enrolled', {course})
        else:
            self.courses.append(course)
    def get_courses(self):
        return self.courses

s1 = Student("james bond", 7)
s2 = Student("clark kent", 333)
s3 = Student("John Doe", 1111)

s1.enroll("BTE320")
s1.enroll("BTE420")

s1.get_courses()
```

```python
#exception handling

def getValues(n):
    L = []
    for i in range(n):
        try:
            x = float(input("Enter a value: "))
            L.append(x)
        except ValueError:
            print("Invalid input. Please enter a number.")
            return []
    return L

def calcAverage(alist):
    if len(alist) == 0:
        print("No values were entered. Cannot calculate average.")
        return None
    sum = 0
    for x in alist:
        sum += x
    return sum / len(alist)

if __name__ == "__main__":
    try:
        n = int(input("Enter the number of values: "))
        L = getValues(n)
        avg = calcAverage(L)
        if avg is not None:
            print(f"The average is: {avg:.2f}")
    except ValueError:
        print("Invalid input. Please enter a whole number.")
```

```python
import math

class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def distance(self, other):
        return math.sqrt((self.x - other.x)**2 + (self.y - other.y)**2)

class Rectangle:
    def __init__(self, corner_1, corner_2, corner_3, corner_4):
        self.corner_1 = corner_1
        self.corner_2 = corner_2
        self.corner_3 = corner_3
        self.corner_4 = corner_4

    def length(self):
        return self.corner_1.distance(self.corner_2)

    def width(self):
        return self.corner_1.distance(self.corner_3)
```
